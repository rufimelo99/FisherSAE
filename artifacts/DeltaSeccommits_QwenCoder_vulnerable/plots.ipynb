```python
import matplotlib.pyplot as plt

# Dataset sizes
dataset_sizes = [1, 5, 10, 25, 50, 100]  # in millions

# Loss metrics (log scale)
mse = [381.726462097168, 0.7027239519357681, 0.5077916109561921, 0.3804915753006935, 0.23148517519235612, 0.15090280279517174]
l0 = [8534.926, 7047.765, 6488.224, 5345.384, 4706.903, 2440.645]
l1 = [7129.927, 24796.585, 21338.863, 9589.514, 5499.656, 3777.367]

# Similarity metrics (linear scale)
l2_ratios = [0.2514629924297333, 0.5459833085536957, 0.6493280667066574, 0.7302418112754822, 0.8078574621677399, 0.8599849504232406]
cos_sim = [0.5112610194087028, 0.6823916840553284, 0.7457929328613281, 0.8072749328613281, 0.8893880158662796, 0.9292638605833053]

# Create subplots side by side
fig, axs = plt.subplots(1, 2, figsize=(16, 6))

# Left: Losses (log scale)
axs[0].plot(dataset_sizes, mse, marker='s', color='red', linewidth=2, label='MSE')
axs[0].plot(dataset_sizes, l0, marker='^', color='purple', linewidth=2, label='L0 Loss')
axs[0].plot(dataset_sizes, l1, marker='o', color='orange', linewidth=2, label='L1 Loss')
axs[0].set_yscale('log')
axs[0].set_xticks(dataset_sizes)
axs[0].set_xticklabels([f"{s}M" for s in dataset_sizes])
axs[0].set_xlabel("Training Tokens (Millions)")
axs[0].set_ylabel("Loss (Log Scale)")
axs[0].set_title("Loss Metrics vs Training Tokens")
axs[0].grid(True, which="both", linestyle='--', alpha=0.5)
axs[0].legend()

# Right: Similarity metrics (linear scale)
axs[1].plot(dataset_sizes, l2_ratios, marker='o', color='blue', linewidth=2, label='L2 Norm Ratio')
axs[1].plot(dataset_sizes, cos_sim, marker='^', color='green', linewidth=2, label='Cosine Similarity')
axs[1].set_xticks(dataset_sizes)
axs[1].set_xticklabels([f"{s}M" for s in dataset_sizes])
axs[1].set_xlabel("Training Tokens (Millions)")
axs[1].set_ylabel("Similarity")
axs[1].set_title("L2 Ratio & Cosine Similarity vs Training Tokens")
axs[1].grid(True, linestyle='--', alpha=0.5)
axs[1].legend()

plt.tight_layout()
#save to pdf
plt.savefig("token_scaling_metrics.pdf")
plt.show()
```

```python
import pandas as pd
loss_mse_path = "mse_loss.csv"
test_mse_loss_path = "test_mse.csv"
df_mse_loss = pd.read_csv(loss_mse_path)
df_test_mse_loss = pd.read_csv(test_mse_loss_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# -------------------------
# NeurIPS-style formatting
# -------------------------
mpl.rcParams.update({
    "font.family": "serif",
    "font.size": 14,
    "axes.titlesize": 16,
    "axes.labelsize": 15,
    "legend.fontsize": 11,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "axes.linewidth": 1.0,
    "pdf.fonttype": 42,   # better for submission
    "ps.fonttype": 42,
})

# Token configs (ordered smallest → largest)
configs = [
    ("1M", "1.000e+06"),
    ("5M", "5.000e+06"),
    ("10M", "1.000e+07"),
    ("25M", "2.500e+07"),
    ("50M", "5.000e+07"),
    ("100M", "1.000e+08"),
]

# Professional color palette (colorblind-friendly)
colors = plt.cm.viridis(np.linspace(0.15, 0.9, len(configs)))

fig, ax = plt.subplots(figsize=(10, 6))

previous_last_step = None

for (label, token_str), color in zip(configs, colors):

    train_col = f"standard-16384-LR-3e-05-Tokens-{token_str} - losses/mse_loss"
    test_col = f"standard-16384-LR-3e-05-Tokens-{token_str} - test/mse"

    train_series = df_mse_loss[train_col].copy()
    test_series = df_test_mse_loss[test_col].copy()

    if previous_last_step is not None:
        train_series[df_mse_loss["Step"] < previous_last_step] = np.nan
        test_series[df_test_mse_loss["Step"] < previous_last_step] = np.nan

    # Train: thin + transparent
    ax.plot(
        df_mse_loss["Step"],
        train_series,
        linestyle="--",
        linewidth=1.8,
        alpha=0.5,
        color=color,
        label=f"{label} Train",
    )

    # Test: thicker + solid (primary signal)
    ax.plot(
        df_test_mse_loss["Step"],
        test_series,
        linestyle="-",
        linewidth=2.8,
        color=color,
        label=f"{label} Test",
    )

    last_valid_idx = df_mse_loss[train_col].last_valid_index()
    if last_valid_idx is not None:
        previous_last_step = df_mse_loss.loc[last_valid_idx, "Step"]

# Axis styling
ax.set_yscale("log")
ax.set_xlabel("Training Steps")
ax.set_ylabel("MSE Loss (log scale)")
ax.set_title("Training vs. Test Loss Across Token Budgets")

# Light grid (subtle)
ax.grid(True, which="both", linestyle="--", linewidth=0.6, alpha=0.3)

# Remove top/right spines for cleaner look
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Compact legend outside
ax.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=False,
    ncol=1
)

plt.tight_layout()
plt.savefig("loss_curve.pdf", bbox_inches="tight")
plt.show()
```

```python
# Identify steps where test MSE loss did NOT drop by more than 1% vs. previous step

configs = [
    ("1M",   "1.000e+06"),
    ("5M",   "5.000e+06"),
    ("10M",  "1.000e+07"),
    ("25M",  "2.500e+07"),
    ("50M",  "5.000e+07"),
    ("100M", "1.000e+08"),
]

THRESHOLD = 0.01  # 1%

stagnation_report = {}

for label, token_str in configs:
    col = f"standard-16384-LR-3e-05-Tokens-{token_str} - test/mse"
    if col not in df_test_mse_loss.columns:
        continue

    series = df_test_mse_loss[col].dropna()
    steps  = df_test_mse_loss.loc[series.index, "Step"]

    # pct_change is (current - previous) / |previous|; drops are negative
    pct_change = series.pct_change()

    # Stagnant = loss didn't fall by more than 1%  (change > -0.01)
    stagnant_mask = pct_change > -THRESHOLD
    stagnant_mask.iloc[0] = False          # first step has no predecessor

    stagnant_steps = steps[stagnant_mask].values
    stagnant_vals  = series[stagnant_mask].values
    stagnant_chg   = pct_change[stagnant_mask].values * 100  # in %

    stagnation_report[label] = list(zip(stagnant_steps, stagnant_vals, stagnant_chg))

    print(f"\n{'='*60}")
    print(f"  Config: {label}  ({len(stagnant_steps)} stagnant steps out of {len(series)-1})")
    print(f"{'='*60}")
    if len(stagnant_steps) == 0:
        print("  None — loss dropped >1% at every step.")
    else:
        print(f"  {'Step':>8}  {'Test MSE':>14}  {'Δ%':>10}")
        for s, v, d in zip(stagnant_steps, stagnant_vals, stagnant_chg):
            print(f"  {s:>8}  {v:>14.4f}  {d:>+10.3f}%")
```
